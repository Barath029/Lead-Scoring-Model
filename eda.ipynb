# CRM Lead Scoring EDA

This notebook checks class balance, source and industry conversion rates, engagement differences, correlations, and time-to-convert. The charts are designed to translate directly into sales and marketing decisions.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv('data/leads_raw.csv', parse_dates=['lead_created_date', 'conversion_date'])
df.head()
```

```python
print(df['converted'].value_counts(normalize=True))
display(df.groupby('lead_source')['converted'].agg(['mean', 'count']).sort_values('mean', ascending=False))
display(df.groupby('industry')['converted'].mean().sort_values(ascending=False))
```

```python
engagement = ['website_visits', 'pages_viewed', 'emails_clicked', 'content_downloads', 'demo_requested', 'meetings_booked']
df.groupby('converted')[engagement].mean().T.plot(kind='bar', figsize=(10, 5), title='Average engagement by conversion outcome')
plt.tight_layout()
```

```python
numeric = df.select_dtypes('number')
plt.figure(figsize=(11, 8))
sns.heatmap(numeric.corr(), cmap='vlag', center=0)
plt.title('Numeric feature correlations')
plt.show()
```

```python
converted = df[df['converted'].eq(1)].copy()
converted['time_to_convert_days'] = (converted['conversion_date'] - converted['lead_created_date']).dt.days
converted['time_to_convert_days'].hist(bins=30, figsize=(9, 4))
plt.title('Time to convert (days)')
plt.xlabel('Days')
plt.show()
```

## Business interpretation

Referral and webinar leads typically show stronger intent signals, while demo requests and meetings are particularly useful prioritization inputs. The correlation plot confirms that engagement measures overlap, so regularization helps avoid over-counting the same behavior. Conversion delay supports a 90-day outcome window, but real CRM snapshots must freeze features at the scoring date.
